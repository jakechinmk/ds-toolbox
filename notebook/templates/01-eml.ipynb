# Setup

## Import Library

```python
import pandas as pd
from pycaret.classification import ClassificationExperiment
```

## Global Path

```python
DATA_DIR = '/home/jake/code/notebook/kaggle/titanic/data/'
TRAIN_PATH = f'{DATA_DIR}train.csv'
TEST_PATH = f'{DATA_DIR}test.csv'

OUTPUT_DIR = f'/home/jake/code/notebook/kaggle/titanic/output/submission/eml/'
```

## Global Parameters

```python
SEED = 123
TARGET_COL = 'survived'

FEATURE_LIST = [
    'pclass',
    'sex',
    'age',
    'sib_sp',
    'parch',
    'fare',
    'embarked'
]

IGNORE_FEATURE_LIST = [
    'passengerid',
    'name',
    'ticket',
    'cabin'
]
```

# Data

```python
df_train = pd.read_csv(TRAIN_PATH)
df_test = pd.read_csv(TEST_PATH)

df_train.columns = [col.lower() for col in df_train.columns]
df_test.columns = [col.lower() for col in df_test.columns]
```

# Exploratory Machine Learning

```python
exp = ClassificationExperiment()
exp.setup(
    data=df_train,
    target=TARGET_COL,
    ignore_features=IGNORE_FEATURE_LIST,
    session_id=SEED,
    use_gpu=True,
)
```

```python
exp.models()
```

```python
best = exp.compare_models(include=['lr', 'rf', 'lightgbm', 'nb'])
```

```python
exp.evaluate_model(best)
```

```python
exp.interpret_model(best)
```

```python
df_pred = exp.predict_model(best, data=df_test.drop(columns=IGNORE_FEATURE_LIST), probability_threshold=0.5) # using KS Stats
df_pred.loc[:, 'PassengerId'] = df_test.passengerid
df_pred.loc[:, 'Survived'] = df_pred.loc[:, 'prediction_label']

SUBMISSION_COL_LIST = ['PassengerId', 'Survived']
df_pred = df_pred.loc[:, SUBMISSION_COL_LIST]
df_pred.to_csv(f'{OUTPUT_DIR}submission.csv', index=False)
```

```python
final_best = exp.finalize_model(best)
df_pred = exp.predict_model(final_best, data=df_test.drop(columns=IGNORE_FEATURE_LIST), probability_threshold=0.5) # using KS Stats
df_pred.loc[:, 'PassengerId'] = df_test.passengerid
df_pred.loc[:, 'Survived'] = df_pred.loc[:, 'prediction_label']

SUBMISSION_COL_LIST = ['PassengerId', 'Survived']
df_pred = df_pred.loc[:, SUBMISSION_COL_LIST]
df_pred.to_csv(f'{OUTPUT_DIR}submission_full.csv', index=False)
```
